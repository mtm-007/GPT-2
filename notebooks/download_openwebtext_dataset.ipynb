```python
os.makedirs("../data/openwebtext/train", exist_ok=True)

# Save each document as a separate file
for i, item in enumerate(dataset["train"]):
    with open(f"../data/openwebtext/train/doc_{i:08d}.txt", "w", encoding="utf-8") as f:
        f.write(item["text"])
    
    # Progress indicator
    if i % 1000 == 0:
        print(f"Saved {i} documents...")
    
    # Optional: limit number of files (remove this to save all)
    # if i >= 10000:
    #     break

print("Done!")
```
